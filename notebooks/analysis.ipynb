```python
import sys, os
sys.path.append(os.path.abspath(".."))
from utils.helpers import clean_lines, parse_products
from utils.product import Product
import matplotlib.pyplot as plt
```

```python
with open("../data/raw/data.txt", "r")  as f:
    lines = f.readlines()
```

```python
lines = clean_lines(lines)
```

```python
products_data = parse_products(lines)
```

```python
products = []
for item in products_data:
    product = Product(item[0], item[1], item[2])
    products.append(product)

total_sales = 0

for product in products:
    value = product.total_value()
    print(product.name, value)
    total_sales += value    
```

```python
product_names = [product.name for product in products]
product_sales = [product.total_value() for product in products]
plt.bar(product_names, product_sales)
plt.title("total sales per product")
plt.xlabel("product name")
plt.ylabel("total sales")
```
